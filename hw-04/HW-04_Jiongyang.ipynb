## 2.2 Binary perceptron classifier

Copy your perceptron class to the cell below. 

* DO NOT use the one from statsmodel. We want to test the perceptron you built.
* Note that your predict method should output `+1` or `-1` for positive or negative values, respectively.
* A functional perceptron classifier gets **4 pt**.

```python
# copy your perceptron class to his cell
import numpy as np

class Perceptron:
    def __init__(self, learning_rate=0.01, n_iterations=1000):
        self.learning_rate = learning_rate
        self.n_iterations = n_iterations
        self.weights = None
        self.bias = None
        
    def fit(self, X, y):
        n_samples, n_features = X.shape
        
        self.weights = np.zeros(n_features)
        self.bias = 0
        
        # Training loop
        for _ in range(self.n_iterations):
            for idx, x_i in enumerate(X):
                # Calculate the linear output
                linear_output = np.dot(x_i, self.weights) + self.bias
                # Apply activation function (step function)
                y_pred = np.where(linear_output >= 0, 1, -1)
                
                if y_pred != y[idx]:
                    update = self.learning_rate * y[idx]
                    self.weights += update * x_i
                    self.bias += update
    
    def predict(self, X):
        
        linear_output = np.dot(X, self.weights) + self.bias
        y_pred = np.where(linear_output >= 0, 1, -1)
        return y_pred
    
    def score(self, X, y):
        predictions = self.predict(X)
        return np.mean(predictions == y)
```

### 2.4 Multi-label perceptron classification

* As we know, perceptron is a binary classifier. For multiple-label classification, we can use One-vs-Rest (OvR) Strategy.
* In this case, let's train five individual perceptrons. 
* For each classifier, it treats the current class as "positive" and all others as "negative."
* When classifying a new sample, each classifier gives a "score," and the class with the highest score is chosen.

Copy your perceptron to the code cell below. We need to add a score method, which outputs dot of weights and features, as opposed to the previous binary predict method. The score method should output a signed floating score value, not `+1` or `-1`. This can be done by removing the binary segmenting, i.e., directly outputing the dot value.

* Functioning score() method gets **2 pt**.

```python
# put your modified perceptron class here
import numpy as np
from sklearn.base import BaseEstimator, ClassifierMixin

class Perceptron:
    def __init__(self, learning_rate=0.01, n_iterations=1000):
        self.learning_rate = learning_rate
        self.n_iterations = n_iterations
        self.weights = None
        self.bias = None
    
    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0
        
        for _ in range(self.n_iterations):
            for idx, x_i in enumerate(X):
                linear_output = np.dot(x_i, self.weights) + self.bias
                y_pred = np.where(linear_output >= 0, 1, -1)
                
                if y_pred != y[idx]:
                    update = self.learning_rate * y[idx]
                    self.weights += update * x_i
                    self.bias += update
    
    def predict(self, X):
        linear_output = np.dot(X, self.weights) + self.bias
        return np.where(linear_output >= 0, 1, -1)
    
    def score(self, X):
        """Outputs the signed floating score value (dot product + bias)"""
        return np.dot(X, self.weights) + self.bias


class MultiClassPerceptron(BaseEstimator, ClassifierMixin):
    def __init__(self, learning_rate=0.01, n_iterations=1000):
        self.learning_rate = learning_rate
        self.n_iterations = n_iterations
        self.classifiers = []
        self.classes_ = None
    
    def fit(self, X, y):
        self.classes_ = np.unique(y)
        self.classifiers = []
        
        for class_label in self.classes_:
            # Create binary labels for OvR
            binary_y = np.where(y == class_label, 1, -1)
            
            # Train individual perceptron
            clf = Perceptron(learning_rate=self.learning_rate,
                            n_iterations=self.n_iterations)
            clf.fit(X, binary_y)
            self.classifiers.append(clf)
    
    def predict(self, X):
        # Get scores from all classifiers
        scores = np.array([clf.score(X) for clf in self.classifiers]).T
        # Predict class with highest score
        return self.classes_[np.argmax(scores, axis=1)]
    
    def predict_proba(self, X):
        """Returns normalized score probabilities"""
        scores = np.array([clf.score(X) for clf in self.classifiers]).T
        # Softmax normalization
        exp_scores = np.exp(scores - np.max(scores, axis=1, keepdims=True))
        return exp_scores / np.sum(exp_scores, axis=1, keepdims=True)
```

```python
# put your code here.
import numpy as np
import pandas as pd
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler

df = pd.read_csv('Telecust1.csv')
df_sample = df.iloc[:200] 

X = df_sample.drop('custcat', axis=1)
y = df_sample['custcat']

if y.dtype == 'object':
    y = y.astype('category').cat.codes

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

param_grid = [
    {'kernel': ['linear'], 'C': [0.1, 1, 10]},
    {'kernel': ['rbf'], 'C': [0.1, 1, 10], 'gamma': ['scale', 'auto']},
    {'kernel': ['poly'], 'C': [0.1, 1], 'degree': [2, 3], 'gamma': ['scale']}
]

print("Performing GridSearchCV (this may take several minutes)...")
svm = SVC()
grid_search = GridSearchCV(svm, param_grid, cv=3, n_jobs=-1, verbose=2)
grid_search.fit(X_train_scaled, y_train)

print("\nBest parameters found:")
print(grid_search.best_params_)

print("\nTraining final model on full dataset...")
full_X = df.drop('custcat', axis=1)
full_y = df['custcat']
if full_y.dtype == 'object':
    full_y = full_y.astype('category').cat.codes

X_train_full, X_test_full, y_train_full, y_test_full = train_test_split(
    full_X, full_y, test_size=0.3, random_state=42)
X_train_full_scaled = scaler.fit_transform(X_train_full)
X_test_full_scaled = scaler.transform(X_test_full)

final_svm = SVC(**grid_search.best_params_)
final_svm.fit(X_train_full_scaled, y_train_full)

train_acc = final_svm.score(X_train_full_scaled, y_train_full)
test_acc = final_svm.score(X_test_full_scaled, y_test_full)

print(f"\nFinal Model Performance:")
print(f"Training Accuracy: {train_acc:.4f}")
print(f"Test Accuracy: {test_acc:.4f}")
```
